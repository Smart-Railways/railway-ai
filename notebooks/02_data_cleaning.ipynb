```python
import pandas as pd
import numpy as np

print("Libraries loaded successfully")
```

```python
assets = pd.read_csv("../data/raw/assets.csv")
maintenance = pd.read_csv("../data/raw/maintenance.csv")
failures = pd.read_csv("../data/raw/failures.csv")
trains = pd.read_csv("../data/raw/trains.csv")
freight = pd.read_csv("../data/raw/freight.csv")

print("Assets:", assets.shape)
print("Maintenance:", maintenance.shape)
print("Failures:", failures.shape)
print("Trains:", trains.shape)
print("Freight:", freight.shape)
```

```python
print("ASSETS")
print(assets.isnull().sum())

print("\nMAINTENANCE")
print(maintenance.isnull().sum())

print("\nFAILURES")
print(failures.isnull().sum())

print("\nTRAINS")
print(trains.isnull().sum())

print("\nFREIGHT")
print(freight.isnull().sum())
```

```python
print("Duplicate assets:", assets.duplicated().sum())
print("Duplicate maintenance:", maintenance.duplicated().sum())
print("Duplicate failures:", failures.duplicated().sum())
print("Duplicate trains:", trains.duplicated().sum())
print("Duplicate freight:", freight.duplicated().sum())
```

```python
print("ASSETS")
print(assets.dtypes)

print("\nMAINTENANCE")
print(maintenance.dtypes)

print("\nFAILURES")
print(failures.dtypes)

print("\nTRAINS")
print(trains.dtypes)

print("\nFREIGHT")
print(freight.dtypes)
```

```python
assets["last_maintenance_date"] = pd.to_datetime(
    assets["last_maintenance_date"]
)

maintenance["created_date"] = pd.to_datetime(
    maintenance["created_date"]
)

maintenance["due_date"] = pd.to_datetime(
    maintenance["due_date"]
)

failures["failure_date"] = pd.to_datetime(
    failures["failure_date"]
)

trains["entry_time"] = pd.to_datetime(
    trains["entry_time"]
)

trains["exit_time"] = pd.to_datetime(
    trains["exit_time"]
)

freight["date"] = pd.to_datetime(
    freight["date"]
)

print("All date columns converted successfully.")
```

```python
print(assets["last_maintenance_date"].dtype)
print(maintenance["due_date"].dtype)
print(failures["failure_date"].dtype)
print(trains["entry_time"].dtype)
print(freight["date"].dtype)
```

```python
print("===== ASSET VALUE CHECKS =====")

print("Criticality outside 1-10:",
      ((assets["criticality"] < 1) | (assets["criticality"] > 10)).sum())

print("Condition score outside 0-100:",
      ((assets["condition_score"] < 0) | (assets["condition_score"] > 100)).sum())

print("Negative failure counts:",
      (assets["failure_count"] < 0).sum())


print("\n===== MAINTENANCE VALUE CHECKS =====")

print("Severity outside 1-10:",
      ((maintenance["severity"] < 1) | (maintenance["severity"] > 10)).sum())

print("Criticality outside 1-10:",
      ((maintenance["criticality"] < 1) | (maintenance["criticality"] > 10)).sum())

print("Negative overdue days:",
      (maintenance["overdue_days"] < 0).sum())

print("Non-positive duration:",
      (maintenance["estimated_duration"] <= 0).sum())

print("Non-positive manpower:",
      (maintenance["required_manpower"] <= 0).sum())


print("\n===== FAILURE VALUE CHECKS =====")

print("Severity outside 1-10:",
      ((failures["severity"] < 1) | (failures["severity"] > 10)).sum())

print("Negative downtime:",
      (failures["downtime_hours"] < 0).sum())

print("Negative repair duration:",
      (failures["repair_duration_hours"] < 0).sum())
```

```python
print("===== MAINTENANCE DATE CHECKS =====")

print(
    "Due date before created date:",
    (maintenance["due_date"] < maintenance["created_date"]).sum()
)

print("\n===== TRAIN TIME CHECKS =====")

print(
    "Exit time before entry time:",
    (trains["exit_time"] < trains["entry_time"]).sum()
)

print("\n===== FAILURE DATE CHECK =====")

print(
    "Future failure dates:",
    (failures["failure_date"] > pd.Timestamp("2026-09-03")).sum()
)

print("\n===== ASSET MAINTENANCE DATE CHECK =====")

print(
    "Future maintenance dates:",
    (assets["last_maintenance_date"] > pd.Timestamp("2026-09-03")).sum()
)
```

```python
print("===== CONDITION SCORE CHECK =====")

print(
    assets[
        (assets["condition_score"] < 0) |
        (assets["condition_score"] > 100)
    ]
)

print("\n===== CRITICALITY CHECK =====")

print(
    assets[
        (assets["criticality"] < 1) |
        (assets["criticality"] > 10)
    ]
)

print("\n===== MAINTENANCE SEVERITY CHECK =====")

print(
    maintenance[
        (maintenance["severity"] < 1) |
        (maintenance["severity"] > 10)
    ]
)
```

```python
print("===== ASSET ID RELATIONSHIP CHECK =====")

invalid_assets = maintenance[
    ~maintenance["asset_id"].isin(assets["asset_id"])
]

print(invalid_assets)


print("\n===== SECTION ID RELATIONSHIP CHECK =====")

invalid_sections = maintenance[
    ~maintenance["section_id"].isin(assets["section_id"])
]

print(invalid_sections)
```

```python
current_date = pd.Timestamp("2026-09-03")

maintenance["calculated_overdue_days"] = (
    current_date - maintenance["due_date"]
).dt.days

maintenance["calculated_overdue_days"] = (
    maintenance["calculated_overdue_days"].clip(lower=0)
)

print(
    maintenance[
        [
            "task_id",
            "due_date",
            "overdue_days",
            "calculated_overdue_days"
        ]
    ]
)
```

```python
assets.to_csv(
    "../data/processed/assets_clean.csv",
    index=False
)

maintenance.to_csv(
    "../data/processed/maintenance_clean.csv",
    index=False
)

failures.to_csv(
    "../data/processed/failures_clean.csv",
    index=False
)

trains.to_csv(
    "../data/processed/trains_clean.csv",
    index=False
)

freight.to_csv(
    "../data/processed/freight_clean.csv",
    index=False
)

print("Cleaned datasets saved successfully.")
```
