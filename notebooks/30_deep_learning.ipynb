```python
import pandas as pd
import numpy as np

history = pd.read_csv(
    "../data/processed/asset_history_simulated.csv"
)

history["snapshot_date"] = pd.to_datetime(
    history["snapshot_date"],
    errors="coerce"
)

history = history.sort_values(
    ["asset_id", "snapshot_date"]
).reset_index(drop=True)

print("Rows:", len(history))
print("Columns:", history.columns.tolist())
print(
    "Date range:",
    history["snapshot_date"].min(),
    "to",
    history["snapshot_date"].max()
)
```

```python
# Step 30.6 FIX — Load the leakage-safe ML history

history = pd.read_csv(
    "../data/processed/failure_ml_dataset.csv"
)

history["snapshot_date"] = pd.to_datetime(
    history["snapshot_date"],
    errors="coerce"
)

history = history.sort_values(
    ["asset_id", "snapshot_date"]
).reset_index(drop=True)

print("Rows:", len(history))
print("Columns:", history.columns.tolist())
print(
    "Date range:",
    history["snapshot_date"].min(),
    "to",
    history["snapshot_date"].max()
)

required_columns = [
    "asset_id",
    "snapshot_date",
    "condition_score",
    "criticality",
    "usage_factor",
    "weather_stress",
    "historical_failure_count",
    "historical_downtime_hours",
    "days_since_last_failure",
    "failure_within_30_days"
]

missing_columns = [
    col for col in required_columns
    if col not in history.columns
]

print("Missing required columns:", missing_columns)
```

```python
SEQUENCE_LENGTH = 12

X_sequences = []
y_sequences = []
sequence_asset_ids = []
sequence_dates = []

for asset_id, group in history.groupby("asset_id"):

    group = group.sort_values("snapshot_date").reset_index(drop=True)

    # Remove rows with missing feature values
    group = group.dropna(
        subset=sequence_features + ["failure_within_30_days"]
    ).reset_index(drop=True)

    for i in range(SEQUENCE_LENGTH, len(group)):

        sequence = group.loc[
            i - SEQUENCE_LENGTH:i - 1,
            sequence_features
        ].values

        target = group.loc[
            i,
            "failure_within_30_days"
        ]

        X_sequences.append(sequence)
        y_sequences.append(target)

        sequence_asset_ids.append(asset_id)
        sequence_dates.append(
            group.loc[i, "snapshot_date"]
        )

X_sequences = np.array(X_sequences, dtype=np.float32)
y_sequences = np.array(y_sequences, dtype=np.float32)

print("X shape:", X_sequences.shape)
print("y shape:", y_sequences.shape)
print("Number of sequences:", len(X_sequences))
print("Positive targets:", int(y_sequences.sum()))
print("Negative targets:", int((y_sequences == 0).sum()))
```

```python
# Step 30.7 — Chronological split

sequence_dates = pd.to_datetime(sequence_dates)

# Convert dates to NumPy array
sequence_dates = np.array(sequence_dates)

# Define time boundaries
TRAIN_END = pd.Timestamp("2024-12-31")
VALIDATION_END = pd.Timestamp("2025-12-31")

# Create masks
train_mask = sequence_dates <= TRAIN_END

validation_mask = (
    (sequence_dates > TRAIN_END)
    & (sequence_dates <= VALIDATION_END)
)

test_mask = sequence_dates > VALIDATION_END

# Split sequences
X_train_dl = X_sequences[train_mask]
y_train_dl = y_sequences[train_mask]

X_validation_dl = X_sequences[validation_mask]
y_validation_dl = y_sequences[validation_mask]

X_test_dl = X_sequences[test_mask]
y_test_dl = y_sequences[test_mask]

# Split dates
dates_train_dl = sequence_dates[train_mask]
dates_validation_dl = sequence_dates[validation_mask]
dates_test_dl = sequence_dates[test_mask]

print("TRAIN")
print("X:", X_train_dl.shape)
print("y:", y_train_dl.shape)
print(
    "Date range:",
    dates_train_dl.min(),
    "to",
    dates_train_dl.max()
)

print("\nVALIDATION")
print("X:", X_validation_dl.shape)
print("y:", y_validation_dl.shape)
print(
    "Date range:",
    dates_validation_dl.min(),
    "to",
    dates_validation_dl.max()
)

print("\nTEST")
print("X:", X_test_dl.shape)
print("y:", y_test_dl.shape)
print(
    "Date range:",
    dates_test_dl.min(),
    "to",
    dates_test_dl.max()
)

print("\nPositive targets")
print("Train:", int(y_train_dl.sum()))
print("Validation:", int(y_validation_dl.sum()))
print("Test:", int(y_test_dl.sum()))
```

```python
# Step 30.8 — Scale sequence features

from sklearn.preprocessing import StandardScaler

# Get dimensions
n_train = X_train_dl.shape
n_validation = X_validation_dl.shape
n_test = X_test_dl.shape

# Create scaler
dl_scaler = StandardScaler()

# Reshape training data:
# (samples, 12 months, 8 features)
# → (samples × 12, 8)
X_train_2d = X_train_dl.reshape(
    -1,
    X_train_dl.shape[-1]
)

# Fit scaler ONLY on training data
dl_scaler.fit(X_train_2d)

# Transform all datasets
X_train_dl_scaled = dl_scaler.transform(
    X_train_2d
).reshape(n_train)

X_validation_dl_scaled = dl_scaler.transform(
    X_validation_dl.reshape(
        -1,
        X_validation_dl.shape[-1]
    )
).reshape(n_validation)

X_test_dl_scaled = dl_scaler.transform(
    X_test_dl.reshape(
        -1,
        X_test_dl.shape[-1]
    )
).reshape(n_test)

print("Original training shape:", X_train_dl.shape)
print("Scaled training shape:", X_train_dl_scaled.shape)

print(
    "Training mean:",
    np.mean(X_train_dl_scaled)
)

print(
    "Training standard deviation:",
    np.std(X_train_dl_scaled)
)

print(
    "Validation shape:",
    X_validation_dl_scaled.shape
)

print(
    "Test shape:",
    X_test_dl_scaled.shape
)
```

```python
# Step 30.9 — Baseline LSTM model

import torch
import torch.nn as nn


class RailwayLSTM(nn.Module):

    def __init__(
        self,
        input_size=8,
        hidden_size=64,
        dense_size=32,
        dropout=0.30
    ):
        super().__init__()

        # LSTM processes the 12-month sequence
        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=1,
            batch_first=True
        )

        # Dropout helps reduce overfitting
        self.dropout1 = nn.Dropout(dropout)

        # Fully connected layer
        self.fc1 = nn.Linear(
            hidden_size,
            dense_size
        )

        self.relu = nn.ReLU()

        self.dropout2 = nn.Dropout(0.20)

        # Final binary classification output
        self.output = nn.Linear(
            dense_size,
            1
        )

    def forward(self, x):

        # LSTM output:
        # shape = (batch, sequence_length, hidden_size)
        lstm_output, (hidden, cell) = self.lstm(x)

        # Take the final time step
        last_output = lstm_output[:, -1, :]

        x = self.dropout1(last_output)

        x = self.fc1(x)

        x = self.relu(x)

        x = self.dropout2(x)

        x = self.output(x)

        # Convert output to probability
        return torch.sigmoid(x)


# Create model
lstm_model = RailwayLSTM(
    input_size=X_train_dl_scaled.shape[2],
    hidden_size=64,
    dense_size=32
)

print(lstm_model)
```

```python
from torch.utils.data import TensorDataset, DataLoader

# Convert NumPy arrays to PyTorch tensors
X_train_tensor = torch.tensor(
    X_train_dl_scaled,
    dtype=torch.float32
)

y_train_tensor = torch.tensor(
    y_train_dl,
    dtype=torch.float32
).reshape(-1, 1)

X_validation_tensor = torch.tensor(
    X_validation_dl_scaled,
    dtype=torch.float32
)

y_validation_tensor = torch.tensor(
    y_validation_dl,
    dtype=torch.float32
).reshape(-1, 1)

X_test_tensor = torch.tensor(
    X_test_dl_scaled,
    dtype=torch.float32
)

y_test_tensor = torch.tensor(
    y_test_dl,
    dtype=torch.float32
).reshape(-1, 1)

print("Train X:", X_train_tensor.shape)
print("Train y:", y_train_tensor.shape)

print("Validation X:", X_validation_tensor.shape)
print("Validation y:", y_validation_tensor.shape)

print("Test X:", X_test_tensor.shape)
print("Test y:", y_test_tensor.shape)
```

```python
# Step 30.11 — Calculate class imbalance

positive_count = int(
    y_train_tensor.sum().item()
)

negative_count = int(
    len(y_train_tensor) - positive_count
)

positive_weight = (
    negative_count / max(positive_count, 1)
)

print("Negative samples:", negative_count)
print("Positive samples:", positive_count)
print(
    "Positive class weight:",
    positive_weight
)
```

```python
# Step 30.11.2 — Create weighted loss

device = torch.device("cpu")

print("Training device:", device)

pos_weight = torch.tensor(
    [positive_weight],
    dtype=torch.float32,
    device=device
)

criterion = nn.BCEWithLogitsLoss(
    pos_weight=pos_weight
)

print("Loss function:", criterion)
```

```python
class RailwayLSTM(nn.Module):

    def __init__(
        self,
        input_size=8,
        hidden_size=64,
        dense_size=32,
        dropout=0.30
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=1,
            batch_first=True
        )

        self.dropout1 = nn.Dropout(dropout)

        self.fc1 = nn.Linear(
            hidden_size,
            dense_size
        )

        self.relu = nn.ReLU()

        self.dropout2 = nn.Dropout(0.20)

        self.output = nn.Linear(
            dense_size,
            1
        )

    def forward(self, x):

        lstm_output, (hidden, cell) = self.lstm(x)

        last_output = lstm_output[:, -1, :]

        x = self.dropout1(last_output)

        x = self.fc1(x)

        x = self.relu(x)

        x = self.dropout2(x)

        return self.output(x)
```

```python
lstm_model = RailwayLSTM(
    input_size=X_train_dl_scaled.shape[2],
    hidden_size=64,
    dense_size=32
)

lstm_model = lstm_model.to(device)

print(lstm_model)
```

```python
# Step 30.12.1 — Optimizer

optimizer = torch.optim.Adam(
    lstm_model.parameters(),
    lr=0.001,
    weight_decay=1e-5
)

print("Optimizer:", optimizer)
```

```python
# Step 30.12.2 — Training and validation functions

def train_one_epoch(
    model,
    loader,
    criterion,
    optimizer,
    device
):

    model.train()

    total_loss = 0.0

    for X_batch, y_batch in loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        logits = model(X_batch)

        loss = criterion(
            logits,
            y_batch
        )

        loss.backward()

        optimizer.step()

        total_loss += (
            loss.item() * X_batch.size(0)
        )

    average_loss = (
        total_loss / len(loader.dataset)
    )

    return average_loss


def evaluate_loss(
    model,
    loader,
    criterion,
    device
):

    model.eval()

    total_loss = 0.0

    with torch.no_grad():

        for X_batch, y_batch in loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            logits = model(X_batch)

            loss = criterion(
                logits,
                y_batch
            )

            total_loss += (
                loss.item() * X_batch.size(0)
            )

    average_loss = (
        total_loss / len(loader.dataset)
    )

    return average_loss
```

```python
# Step 30.12.3 — Training configuration

NUM_EPOCHS = 30

PATIENCE = 5

best_validation_loss = float("inf")

epochs_without_improvement = 0

training_history = []

best_model_path = "../models/best_lstm_failure_model.pt"

print("Maximum epochs:", NUM_EPOCHS)
print("Early stopping patience:", PATIENCE)
print("Best model path:", best_model_path)
```

```python
# Step 30.12.4 — Train LSTM

for epoch in range(1, NUM_EPOCHS + 1):

    train_loss = train_one_epoch(
        lstm_model,
        train_loader,
        criterion,
        optimizer,
        device
    )

    validation_loss = evaluate_loss(
        lstm_model,
        validation_loader,
        criterion,
        device
    )

    training_history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "validation_loss": validation_loss
    })

    print(
        f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
        f"Train Loss: {train_loss:.6f} | "
        f"Validation Loss: {validation_loss:.6f}"
    )

    # Check whether validation improved
    if validation_loss < best_validation_loss:

        best_validation_loss = validation_loss

        epochs_without_improvement = 0

        torch.save(
            lstm_model.state_dict(),
            best_model_path
        )

        print("  → Best model saved.")

    else:

        epochs_without_improvement += 1

        print(
            f"  → No improvement "
            f"({epochs_without_improvement}/{PATIENCE})"
        )

    # Early stopping
    if epochs_without_improvement >= PATIENCE:

        print(
            "\nEarly stopping triggered."
        )

        break
```

```python
# Step 30.12.5 — Restore best LSTM

lstm_model.load_state_dict(
    torch.load(
        best_model_path,
        map_location=device
    )
)

lstm_model.eval()

print(
    "Best LSTM model restored."
)

print(
    "Best validation loss:",
    best_validation_loss
)
```

```python
# Step 30.13.1 — Generate LSTM predictions

lstm_model.eval()

def predict_probabilities(
    model,
    loader,
    device
):
    
    probabilities = []
    actual_values = []

    with torch.no_grad():

        for X_batch, y_batch in loader:

            X_batch = X_batch.to(device)

            logits = model(X_batch)

            probs = torch.sigmoid(logits)

            probabilities.extend(
                probs.cpu().numpy().ravel()
            )

            actual_values.extend(
                y_batch.numpy().ravel()
            )

    return (
        np.array(probabilities),
        np.array(actual_values)
    )


lstm_validation_probabilities, lstm_validation_actual = (
    predict_probabilities(
        lstm_model,
        validation_loader,
        device
    )
)

lstm_test_probabilities, lstm_test_actual = (
    predict_probabilities(
        lstm_model,
        test_loader,
        device
    )
)

print(
    "Validation predictions:",
    len(lstm_validation_probabilities)
)

print(
    "Test predictions:",
    len(lstm_test_probabilities)
)
```

```python
# Step 30.13.2 — Inspect probabilities

print(
    "Minimum probability:",
    lstm_test_probabilities.min()
)

print(
    "Maximum probability:",
    lstm_test_probabilities.max()
)

print(
    "Mean probability:",
    lstm_test_probabilities.mean()
)

print(
    "Median probability:",
    np.median(lstm_test_probabilities)
)
```

```python
# Step 30.13.3 — Classification at threshold 0.5

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

LSTM_THRESHOLD = 0.50

lstm_test_predictions = (
    lstm_test_probabilities >= LSTM_THRESHOLD
).astype(int)

lstm_accuracy = accuracy_score(
    lstm_test_actual,
    lstm_test_predictions
)

lstm_precision = precision_score(
    lstm_test_actual,
    lstm_test_predictions,
    zero_division=0
)

lstm_recall = recall_score(
    lstm_test_actual,
    lstm_test_predictions,
    zero_division=0
)

lstm_f1 = f1_score(
    lstm_test_actual,
    lstm_test_predictions,
    zero_division=0
)

lstm_roc_auc = roc_auc_score(
    lstm_test_actual,
    lstm_test_probabilities
)

lstm_pr_auc = average_precision_score(
    lstm_test_actual,
    lstm_test_probabilities
)

print("LSTM Test Results — Threshold 0.50")
print("----------------------------------")
print("Accuracy :", lstm_accuracy)
print("Precision:", lstm_precision)
print("Recall   :", lstm_recall)
print("F1       :", lstm_f1)
print("ROC-AUC  :", lstm_roc_auc)
print("PR-AUC   :", lstm_pr_auc)
```

```python
# Step 30.16.2 — 1D CNN model

class RailwayCNN(nn.Module):

    def __init__(
        self,
        input_channels=8,
        sequence_length=12,
        dropout=0.30
    ):
        super().__init__()

        self.conv1 = nn.Conv1d(
            in_channels=input_channels,
            out_channels=32,
            kernel_size=3,
            padding=1
        )

        self.bn1 = nn.BatchNorm1d(32)

        self.conv2 = nn.Conv1d(
            in_channels=32,
            out_channels=64,
            kernel_size=3,
            padding=1
        )

        self.bn2 = nn.BatchNorm1d(64)

        self.relu = nn.ReLU()

        self.pool = nn.MaxPool1d(
            kernel_size=2
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.flatten = nn.Flatten()

        self.fc1 = nn.Linear(
            64 * (sequence_length // 2),
            32
        )

        self.output = nn.Linear(
            32,
            1
        )

    def forward(self, x):

        # (batch, 12, 8)
        # → (batch, 8, 12)

        x = x.transpose(1, 2)

        x = self.conv1(x)
        x = self.bn1(x)
        x = self.relu(x)

        x = self.conv2(x)
        x = self.bn2(x)
        x = self.relu(x)

        x = self.pool(x)

        x = self.dropout(x)

        x = self.flatten(x)

        x = self.fc1(x)
        x = self.relu(x)

        x = self.dropout(x)

        # Raw logits
        return self.output(x)
```

```python
# Step 30.18.4 — LSTM Autoencoder

class RailwayAutoencoder(nn.Module):

    def __init__(
        self,
        input_size=8,
        hidden_size=64,
        latent_size=16
    ):
        super().__init__()

        self.encoder = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            batch_first=True
        )

        self.latent = nn.Linear(
            hidden_size,
            latent_size
        )

        self.decoder_input = nn.Linear(
            latent_size,
            hidden_size
        )

        self.decoder = nn.LSTM(
            input_size=hidden_size,
            hidden_size=hidden_size,
            batch_first=True
        )

        self.output = nn.Linear(
            hidden_size,
            input_size
        )

    def forward(self, x):

        # Encode sequence
        encoded, (hidden, cell) = self.encoder(x)

        # Use final timestep
        last_hidden = encoded[:, -1, :]

        # Compress into latent representation
        latent = self.latent(last_hidden)

        # Expand latent representation
        decoder_seed = self.decoder_input(latent)

        # Repeat across all 12 months
        decoder_input = decoder_seed.unsqueeze(1).repeat(
            1,
            x.size(1),
            1
        )

        # Decode
        decoded, _ = self.decoder(
            decoder_input
        )

        reconstructed = self.output(
            decoded
        )

        return reconstructed
```

```python
# Step 30.18.5 — Create autoencoder

autoencoder = RailwayAutoencoder(
    input_size=len(sequence_features),
    hidden_size=64,
    latent_size=16
)

autoencoder = autoencoder.to(device)

print(autoencoder)
```

```python
# Step 30.20.2 — Positional encoding

class PositionalEncoding(nn.Module):

    def __init__(
        self,
        d_model,
        max_len=12
    ):
        super().__init__()

        position = torch.arange(
            max_len
        ).unsqueeze(1)

        div_term = torch.exp(
            torch.arange(
                0,
                d_model,
                2
            ) *
            (
                -np.log(10000.0)
                / d_model
            )
        )

        pe = torch.zeros(
            max_len,
            d_model
        )

        pe[:, 0::2] = torch.sin(
            position * div_term
        )

        pe[:, 1::2] = torch.cos(
            position * div_term
        )

        pe = pe.unsqueeze(0)

        self.register_buffer(
            "pe",
            pe
        )

    def forward(self, x):

        return x + self.pe[
            :, :x.size(1), :
        ]
```

```python
# Step 30.20.3 — Railway Transformer

class RailwayTransformer(nn.Module):

    def __init__(
        self,
        input_size=8,
        d_model=64,
        nhead=4,
        num_layers=2,
        dim_feedforward=128,
        dropout=0.20
    ):
        super().__init__()

        self.input_projection = nn.Linear(
            input_size,
            d_model
        )

        self.position_encoding = PositionalEncoding(
            d_model=d_model,
            max_len=SEQUENCE_LENGTH
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_feedforward,
            dropout=dropout,
            batch_first=True,
            activation="gelu"
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.fc1 = nn.Linear(
            d_model,
            32
        )

        self.relu = nn.ReLU()

        self.output = nn.Linear(
            32,
            1
        )

    def forward(self, x):

        # Convert 8 input features → d_model
        x = self.input_projection(x)

        # Add temporal position information
        x = self.position_encoding(x)

        # Transformer attention
        x = self.transformer(x)

        # Use final timestep representation
        x = x[:, -1, :]

        x = self.dropout(x)

        x = self.fc1(x)

        x = self.relu(x)

        x = self.output(x)

        return torch.sigmoid(x)
```

```python
# Step 30.20.11.1 — Transformer predictions

def get_transformer_probabilities(
    model,
    data_loader,
    device
):

    model.eval()

    probabilities = []

    with torch.no_grad():

        for X_batch, _ in data_loader:

            X_batch = X_batch.to(device)

            logits = model(X_batch)

            batch_probabilities = torch.sigmoid(
                logits
            )

            probabilities.extend(
                batch_probabilities.cpu().numpy().flatten()
            )

    return np.array(probabilities)


transformer_validation_probabilities = (
    get_transformer_probabilities(
        transformer_model,
        validation_loader,
        device
    )
)

transformer_test_probabilities = (
    get_transformer_probabilities(
        transformer_model,
        test_loader,
        device
    )
)

print(
    "Validation predictions:",
    len(transformer_validation_probabilities)
)

print(
    "Test predictions:",
    len(transformer_test_probabilities)
)
```

```python
# Step 30.20.11.3 — Transformer validation threshold analysis

transformer_threshold_results = []

for threshold in np.arange(
    0.01,
    1.00,
    0.01
):

    predictions = (
        transformer_validation_probabilities
        >= threshold
    ).astype(int)

    precision = precision_score(
        y_validation_dl,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_validation_dl,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_validation_dl,
        predictions,
        zero_division=0
    )

    transformer_threshold_results.append({
        "threshold": threshold,
        "precision": precision,
        "recall": recall,
        "f1": f1
    })

transformer_threshold_results_df = pd.DataFrame(
    transformer_threshold_results
)

best_transformer_row = (
    transformer_threshold_results_df
    .sort_values(
        "f1",
        ascending=False
    )
    .iloc[0]
)

best_transformer_threshold = float(
    best_transformer_row["threshold"]
)

print(
    "Best Transformer threshold:",
    best_transformer_threshold
)

print(
    "Validation Precision:",
    best_transformer_row["precision"]
)

print(
    "Validation Recall:",
    best_transformer_row["recall"]
)

print(
    "Validation F1:",
    best_transformer_row["f1"]
)
```

```python
# Step 30.21.7 — Standardize prediction columns

def standardize_prediction_dataframe(
    df,
    model_name
):

    output = df.copy()

    probability_candidates = [
        "failure_probability",
        "probability",
        "predicted_probability",
        "prediction_probability"
    ]

    probability_column = None

    for column in probability_candidates:

        if column in output.columns:
            probability_column = column
            break

    if probability_column is None:

        raise ValueError(
            f"No probability column found for {model_name}. "
            f"Available columns: {output.columns.tolist()}"
        )

    output = output.rename(
        columns={
            probability_column:
            f"{model_name}_probability"
        }
    )

    return output
```

```python
# Step 30.25.1 — Transformer with attention output

class RailwayTransformerWithAttention(nn.Module):

    def __init__(
        self,
        input_size=8,
        d_model=64,
        nhead=4,
        num_layers=2,
        dropout=0.20
    ):
        super().__init__()

        self.input_projection = nn.Linear(
            input_size,
            d_model
        )

        self.positional_encoding = nn.Parameter(
            torch.randn(1, SEQUENCE_LENGTH, d_model)
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=128,
            dropout=dropout,
            batch_first=True
        )

        self.encoder = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.dropout = nn.Dropout(dropout)

        self.output = nn.Linear(
            d_model,
            1
        )

    def forward(
        self,
        x
    ):

        x = self.input_projection(x)

        x = (
            x
            + self.positional_encoding[:, :x.size(1), :]
        )

        encoded = self.encoder(x)

        pooled = encoded.mean(
            dim=1
        )

        pooled = self.dropout(
            pooled
        )

        logits = self.output(
            pooled
        )

        return logits
```

```python
# Step 30.25.4 — Explicit temporal attention model

class TemporalAttentionTransformer(nn.Module):

    def __init__(
        self,
        input_size=8,
        d_model=64,
        nhead=4,
        dropout=0.20
    ):
        super().__init__()

        self.input_projection = nn.Linear(
            input_size,
            d_model
        )

        self.positional_encoding = nn.Parameter(
            torch.randn(
                1,
                SEQUENCE_LENGTH,
                d_model
            )
        )

        self.attention = nn.MultiheadAttention(
            embed_dim=d_model,
            num_heads=nhead,
            dropout=dropout,
            batch_first=True
        )

        self.norm = nn.LayerNorm(
            d_model
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.output = nn.Linear(
            d_model,
            1
        )

    def forward(
        self,
        x,
        return_attention=False
    ):

        x = self.input_projection(x)

        x = (
            x
            + self.positional_encoding[
                :, :x.size(1), :
            ]
        )

        attended, attention_weights = (
            self.attention(
                x,
                x,
                x,
                need_weights=True,
                average_attn_weights=False
            )
        )

        x = self.norm(
            x + self.dropout(attended)
        )

        pooled = x.mean(
            dim=1
        )

        logits = self.output(
            pooled
        )

        if return_attention:

            return (
                logits,
                attention_weights
            )

        return logits
```

```python
# Step 30.25.5 — Initialize temporal attention model

temporal_attention_model = (
    TemporalAttentionTransformer(
        input_size=len(sequence_features),
        d_model=64,
        nhead=4,
        dropout=0.20
    )
)

print(
    temporal_attention_model
)
```

```python
# Step 30.25.7 — Training configuration

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

temporal_attention_model = (
    temporal_attention_model.to(device)
)

criterion = nn.BCEWithLogitsLoss(
    pos_weight=torch.tensor(
        [pos_weight_value],
        dtype=torch.float32,
        device=device
    )
)

optimizer = torch.optim.AdamW(
    temporal_attention_model.parameters(),
    lr=0.001,
    weight_decay=1e-4
)

print(
    "Device:",
    device
)
```

```python
# Step 30.25.8 — Train temporal attention model

EPOCHS = 20

attention_training_history = []

for epoch in range(EPOCHS):

    temporal_attention_model.train()

    total_train_loss = 0.0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)

        y_batch = y_batch.to(device)

        optimizer.zero_grad()

        logits = temporal_attention_model(
            X_batch
        )

        loss = criterion(
            logits,
            y_batch
        )

        loss.backward()

        optimizer.step()

        total_train_loss += (
            loss.item()
            * X_batch.size(0)
        )

    average_train_loss = (
        total_train_loss
        / len(train_loader.dataset)
    )

    # Validation
    temporal_attention_model.eval()

    total_validation_loss = 0.0

    with torch.no_grad():

        for X_batch, y_batch in validation_loader:

            X_batch = X_batch.to(device)

            y_batch = y_batch.to(device)

            logits = temporal_attention_model(
                X_batch
            )

            loss = criterion(
                logits,
                y_batch
            )

            total_validation_loss += (
                loss.item()
                * X_batch.size(0)
            )

    average_validation_loss = (
        total_validation_loss
        / len(validation_loader.dataset)
    )

    attention_training_history.append({
        "epoch": epoch + 1,
        "train_loss": average_train_loss,
        "validation_loss": average_validation_loss
    })

    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Train Loss: {average_train_loss:.4f} | "
        f"Validation Loss: {average_validation_loss:.4f}"
    )
```

```python
class TemporalAttentionTransformer(nn.Module):

    def __init__(
        self,
        input_size=8,
        d_model=64,
        nhead=4,
        dropout=0.20
    ):
        super().__init__()

        self.input_projection = nn.Linear(
            input_size,
            d_model
        )

        self.positional_encoding = nn.Parameter(
            torch.randn(
                1,
                SEQUENCE_LENGTH,
                d_model
            )
        )

        self.attention = nn.MultiheadAttention(
            embed_dim=d_model,
            num_heads=nhead,
            dropout=dropout,
            batch_first=True
        )

        self.norm = nn.LayerNorm(d_model)

        self.dropout = nn.Dropout(dropout)

        self.output = nn.Linear(
            d_model,
            1
        )

    def forward(self, x, return_attention=False):

        # Convert original features to attention dimension
        x = self.input_projection(x)

        # Add temporal positional information
        x = x + self.positional_encoding[:, :x.size(1), :]

        # Self-attention
        attended, attention_weights = self.attention(
            x,
            x,
            x,
            need_weights=True,
            average_attn_weights=False
        )

        # Residual connection + normalization
        x = self.norm(
            x + self.dropout(attended)
        )

        # Aggregate information across 12 months
        pooled = x.mean(dim=1)

        # Failure prediction
        logits = self.output(pooled)

        if return_attention:
            return logits, attention_weights

        return logits
```

```python
# 30.26.1 — Robust prediction function

def predict_dl_model(X):
    """
    Generate failure probabilities from the selected deep-learning model.
    Supports LSTM, CNN, Transformer, and Deep Learning Ensemble.
    """
    
    if selected_dl_model_name == "LSTM":
        model = lstm_model
        model.eval()
        
        with torch.no_grad():
            logits = model(
                torch.tensor(X, dtype=torch.float32).to(device)
            )
            probabilities = torch.sigmoid(logits).cpu().numpy().reshape(-1)

        return probabilities

    elif selected_dl_model_name == "CNN":
        model = cnn_model
        model.eval()
        
        with torch.no_grad():
            logits = model(
                torch.tensor(X, dtype=torch.float32).to(device)
            )
            probabilities = torch.sigmoid(logits).cpu().numpy().reshape(-1)

        return probabilities

    elif selected_dl_model_name == "Transformer":
        model = transformer_model
        model.eval()
        
        with torch.no_grad():
            logits = model(
                torch.tensor(X, dtype=torch.float32).to(device)
            )
            probabilities = torch.sigmoid(logits).cpu().numpy().reshape(-1)

        return probabilities

    elif selected_dl_model_name == "Deep Learning Ensemble":
        predictions = []

        for model in [lstm_model, cnn_model, transformer_model]:
            model.eval()
            
            with torch.no_grad():
                logits = model(
                    torch.tensor(X, dtype=torch.float32).to(device)
                )
                probabilities = torch.sigmoid(logits).cpu().numpy().reshape(-1)
            
            predictions.append(probabilities)

        return np.mean(predictions, axis=0)

    else:
        raise ValueError(
            f"Unknown selected model: {selected_dl_model_name}"
        )


baseline_stress_probabilities = predict_dl_model(
    X_test_dl_scaled
)

print("Selected model:", selected_dl_model_name)
print("Number of predictions:", len(baseline_stress_probabilities))
print(
    "Mean baseline probability:",
    round(baseline_stress_probabilities.mean(), 6)
)
```

```python
# 30.26.2 — Stress-test evaluation

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)

def evaluate_stress_test(
    name,
    probabilities,
    baseline_probabilities,
    y_true
):
    
    probability_change = np.abs(
        probabilities - baseline_probabilities
    )

    baseline_predictions = (
        baseline_probabilities >= 0.5
    ).astype(int)

    stress_predictions = (
        probabilities >= 0.5
    ).astype(int)

    prediction_flip_rate = np.mean(
        baseline_predictions != stress_predictions
    )

    return {
        "Scenario": name,
        "Mean Probability": probabilities.mean(),
        "Mean Absolute Probability Change": probability_change.mean(),
        "Maximum Probability Change": probability_change.max(),
        "Prediction Flip Rate": prediction_flip_rate,
        "ROC_AUC": roc_auc_score(y_true, probabilities),
        "PR_AUC": average_precision_score(y_true, probabilities)
    }
```

```python
# 30.27.1 — Monte Carlo Dropout uncertainty

def enable_dropout(model):
    """
    Enable dropout layers during inference
    while keeping the rest of the model in evaluation mode.
    """
    model.eval()

    for module in model.modules():
        if isinstance(module, nn.Dropout):
            module.train()


def predict_with_mc_dropout(
    model,
    X,
    n_samples=30
):
    """
    Generate multiple predictions using Monte Carlo Dropout.
    """

    model.eval()
    enable_dropout(model)

    X_tensor = torch.tensor(
        X,
        dtype=torch.float32
    ).to(device)

    predictions = []

    with torch.no_grad():

        for _ in range(n_samples):

            logits = model(X_tensor)

            probabilities = torch.sigmoid(
                logits
            ).cpu().numpy().reshape(-1)

            predictions.append(probabilities)

    predictions = np.array(predictions)

    return predictions


print("MC Dropout functions created.")
```

```python
# 30.27.2 — Generate uncertainty samples

if selected_dl_model_name == "LSTM":

    mc_predictions = predict_with_mc_dropout(
        lstm_model,
        X_test_dl_scaled,
        n_samples=30
    )

elif selected_dl_model_name == "CNN":

    mc_predictions = predict_with_mc_dropout(
        cnn_model,
        X_test_dl_scaled,
        n_samples=30
    )

elif selected_dl_model_name == "Transformer":

    mc_predictions = predict_with_mc_dropout(
        transformer_model,
        X_test_dl_scaled,
        n_samples=30
    )

else:

    raise ValueError(
        "MC Dropout uncertainty is implemented "
        "for a single selected deep-learning model."
    )


print(
    "MC prediction shape:",
    mc_predictions.shape
)
```

```python
# 30.27.3 — Calculate prediction uncertainty

mc_mean_probability = mc_predictions.mean(axis=0)

mc_std_probability = mc_predictions.std(axis=0)

mc_lower_probability = np.percentile(
    mc_predictions,
    5,
    axis=0
)

mc_upper_probability = np.percentile(
    mc_predictions,
    95,
    axis=0
)

print(
    "Mean probability:",
    round(mc_mean_probability.mean(), 6)
)

print(
    "Mean uncertainty:",
    round(mc_std_probability.mean(), 6)
)
```

```python
# 30.27.4 — Confidence score

uncertainty_scale = (
    mc_std_probability.max()
    - mc_std_probability.min()
)

if uncertainty_scale == 0:

    confidence_score = np.ones_like(
        mc_std_probability
    )

else:

    confidence_score = (
        1
        - (
            (mc_std_probability - mc_std_probability.min())
            / uncertainty_scale
        )
    )

confidence_score = np.clip(
    confidence_score,
    0,
    1
)

print(
    "Mean confidence:",
    round(confidence_score.mean(), 6)
)
```

```python
# 30.32.1 — Load model evidence

logistic_predictions = pd.read_csv(
    "../data/predictions/"
    "logistic_failure_predictions.csv"
)

tree_predictions = pd.read_csv(
    "../data/predictions/"
    "decision_tree_failure_predictions.csv"
)

rf_predictions = pd.read_csv(
    "../data/predictions/"
    "random_forest_failure_predictions.csv"
)

xgb_predictions = pd.read_csv(
    "../data/predictions/"
    "xgboost_failure_predictions.csv"
)

dl_uncertainty = pd.read_csv(
    "../data/predictions/"
    "deep_learning_uncertainty_predictions.csv"
)

temporal_patterns = pd.read_csv(
    "../data/predictions/"
    "deep_learning_temporal_patterns.csv"
)

print("Logistic:", len(logistic_predictions))
print("Tree:", len(tree_predictions))
print("Random Forest:", len(rf_predictions))
print("XGBoost:", len(xgb_predictions))
print("DL:", len(dl_uncertainty))
print("Temporal:", len(temporal_patterns))
```

```python
# 30.32.2 — Standardize model evidence

def standardize_prediction(
    df,
    probability_column,
    model_name
):
    
    output = df[
        [
            "asset_id",
            "snapshot_date"
        ]
    ].copy()

    output["snapshot_date"] = pd.to_datetime(
        output["snapshot_date"]
    )

    output[model_name] = pd.to_numeric(
        df[probability_column],
        errors="coerce"
    )

    return output


logistic_std = standardize_prediction(
    logistic_predictions,
    "failure_probability",
    "logistic_probability"
)

tree_std = standardize_prediction(
    tree_predictions,
    "failure_probability",
    "tree_probability"
)

rf_std = standardize_prediction(
    rf_predictions,
    "failure_probability",
    "rf_probability"
)

xgb_std = standardize_prediction(
    xgb_predictions,
    "failure_probability",
    "xgb_probability"
)

dl_std = dl_uncertainty[
    [
        "asset_id",
        "snapshot_date",
        "mean_failure_probability",
        "uncertainty_std",
        "confidence_score"
    ]
].copy()

dl_std["snapshot_date"] = pd.to_datetime(
    dl_std["snapshot_date"]
)

dl_std = dl_std.rename(
    columns={
        "mean_failure_probability":
            "dl_probability",
        "uncertainty_std":
            "dl_uncertainty",
        "confidence_score":
            "dl_confidence"
    }
)

print(logistic_std.head())
print(dl_std.head())
```

```python
# 30.32.3 — Merge classical model predictions

evidence_df = logistic_std.merge(
    tree_std,
    on=["asset_id", "snapshot_date"],
    how="outer"
)

evidence_df = evidence_df.merge(
    rf_std,
    on=["asset_id", "snapshot_date"],
    how="outer"
)

evidence_df = evidence_df.merge(
    xgb_std,
    on=["asset_id", "snapshot_date"],
    how="outer"
)

print(
    "Rows after classical merge:",
    len(evidence_df)
)

evidence_df.head()
```

```python
# 30.32.4 — Add deep-learning evidence

evidence_df = evidence_df.merge(
    dl_std,
    on=["asset_id", "snapshot_date"],
    how="outer"
)

print(
    "Rows after DL merge:",
    len(evidence_df)
)

evidence_df.head()
```

```python
# 30.32.5 — Add temporal evidence

temporal_std = temporal_patterns[
    [
        "asset_id",
        "degradation_pattern_score",
        "recent_condition_change",
        "long_term_condition_change"
    ]
].copy()

evidence_df = evidence_df.merge(
    temporal_std,
    on="asset_id",
    how="left"
)

print(
    "Rows after temporal merge:",
    len(evidence_df)
)
```

```python
# 30.32.6 — Missing evidence analysis

evidence_columns = [
    "logistic_probability",
    "tree_probability",
    "rf_probability",
    "xgb_probability",
    "dl_probability",
    "dl_uncertainty",
    "dl_confidence",
    "degradation_pattern_score"
]

missing_evidence = (
    evidence_df[evidence_columns]
    .isna()
    .sum()
    .sort_values(
        ascending=False
    )
)

missing_evidence
```

```python
# 30.32.7 — Complete model evidence

complete_model_mask = (
    evidence_df[
        [
            "logistic_probability",
            "tree_probability",
            "rf_probability",
            "xgb_probability",
            "dl_probability"
        ]
    ]
    .notna()
    .all(axis=1)
)

complete_evidence = evidence_df[
    complete_model_mask
].copy()

print(
    "Complete evidence rows:",
    len(complete_evidence)
)

print(
    "Total evidence rows:",
    len(evidence_df)
)
```

```python
# 30.32.8 — Equal-weight ensemble

model_probability_columns = [
    "logistic_probability",
    "tree_probability",
    "rf_probability",
    "xgb_probability",
    "dl_probability"
]

complete_evidence[
    "ensemble_probability_equal"
] = (
    complete_evidence[
        model_probability_columns
    ]
    .mean(axis=1)
)

complete_evidence[
    [
        "asset_id",
        "snapshot_date",
        "ensemble_probability_equal"
    ]
].head(10)
```

```python
# 30.32.9 — Weighted ensemble

weights = {
    "logistic_probability": 0.20,
    "tree_probability": 0.10,
    "rf_probability": 0.10,
    "xgb_probability": 0.30,
    "dl_probability": 0.30
}

weight_total = sum(weights.values())

assert abs(weight_total - 1.0) < 1e-9

complete_evidence[
    "ensemble_probability_weighted"
] = 0.0

for column, weight in weights.items():

    complete_evidence[
        "ensemble_probability_weighted"
    ] += (
        weight
        * complete_evidence[column]
    )

complete_evidence[
    [
        "asset_id",
        "ensemble_probability_equal",
        "ensemble_probability_weighted"
    ]
].head(10)
```

```python
# 30.32.10 — Confidence-aware ensemble

complete_evidence[
    "confidence_adjustment"
] = (
    0.5
    +
    0.5
    * complete_evidence[
        "dl_confidence"
    ].clip(0, 1)
)

complete_evidence[
    "confidence_aware_probability"
] = (
    complete_evidence[
        "ensemble_probability_weighted"
    ]
    *
    complete_evidence[
        "confidence_adjustment"
    ]
)

complete_evidence[
    [
        "ensemble_probability_weighted",
        "dl_confidence",
        "confidence_adjustment",
        "confidence_aware_probability"
    ]
].head(10)
```

```python
# 30.32.11 — Temporal evidence fusion

complete_evidence[
    "degradation_signal"
] = (
    complete_evidence[
        "degradation_pattern_score"
    ]
    .fillna(0)
    .clip(0, 1)
)

complete_evidence[
    "unified_ai_risk"
] = (
    0.80
    * complete_evidence[
        "confidence_aware_probability"
    ]
    +
    0.20
    * complete_evidence[
        "degradation_signal"
    ]
)

complete_evidence[
    [
        "asset_id",
        "confidence_aware_probability",
        "degradation_signal",
        "unified_ai_risk"
    ]
].head(20)
```

```python
# 30.32.12 — Unified risk categories

complete_evidence[
    "unified_risk_category"
] = pd.cut(
    complete_evidence[
        "unified_ai_risk"
    ],
    bins=[
        -np.inf,
        0.25,
        0.50,
        0.75,
        np.inf
    ],
    labels=[
        "LOW",
        "MEDIUM",
        "HIGH",
        "CRITICAL"
    ]
)

complete_evidence[
    "unified_risk_category"
].value_counts()
```

```python
# 30.32.13 — Ensemble evaluation

xgb_actual = xgb_predictions[
    [
        "asset_id",
        "snapshot_date",
        "failure_within_30_days"
    ]
].copy()

xgb_actual["snapshot_date"] = pd.to_datetime(
    xgb_actual["snapshot_date"]
)

complete_evidence = complete_evidence.merge(
    xgb_actual,
    on=["asset_id", "snapshot_date"],
    how="left"
)

complete_evidence = complete_evidence.dropna(
    subset=["failure_within_30_days"]
)

y_ensemble = (
    complete_evidence[
        "failure_within_30_days"
    ].astype(int)
)

equal_prob = complete_evidence[
    "ensemble_probability_equal"
]

weighted_prob = complete_evidence[
    "ensemble_probability_weighted"
]

unified_prob = complete_evidence[
    "unified_ai_risk"
]

ensemble_results = pd.DataFrame([
    {
        "Model": "Equal Ensemble",
        "ROC_AUC": roc_auc_score(
            y_ensemble,
            equal_prob
        ),
        "PR_AUC": average_precision_score(
            y_ensemble,
            equal_prob
        )
    },
    {
        "Model": "Weighted Ensemble",
        "ROC_AUC": roc_auc_score(
            y_ensemble,
            weighted_prob
        ),
        "PR_AUC": average_precision_score(
            y_ensemble,
            weighted_prob
        )
    },
    {
        "Model": "Unified AI Risk",
        "ROC_AUC": roc_auc_score(
            y_ensemble,
            unified_prob
        ),
        "PR_AUC": average_precision_score(
            y_ensemble,
            unified_prob
        )
    }
])

ensemble_results
```
